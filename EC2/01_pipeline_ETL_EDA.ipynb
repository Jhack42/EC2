Importando pandas

```python
import pandas as pd
```

Importando el dataset

```python
df = pd.read_csv(
    "DATASET_CEXTERNA_2015_2022_0.csv",
    sep=";",
    encoding="utf-8-sig",
    dtype={"UBIGEO": "string"}
)
```

![Descripción de la imagen](https://i.postimg.cc/14xdZ9Jv/Code-q4Zu9Dvhbw.png)

```python
# Corregir caracteres mal codificados
def corregir_texto(texto):
    if isinstance(texto, str):
        try:
            return texto.encode("latin1").decode("utf-8")
        except (UnicodeEncodeError, UnicodeDecodeError):
            return texto
    return texto

# Corregir nombres de columnas
df.columns = [corregir_texto(col) for col in df.columns]

# Corregir valores de texto
for col in df.select_dtypes(include=["object", "string"]).columns:
    df[col] = df[col].map(corregir_texto)

print("Corrección aplicada")
```

Normalizar cadenas vacías y espacios en blanco


```python

columnas_texto = df.select_dtypes(
    include=["object", "string"]
).columns

for col in columnas_texto:
    df[col] = df[col].map(
        lambda x: pd.NA
        if isinstance(x, str) and x.strip() == ""
        else x
    )
```

```python
# Corregir la columna de fecha
df["FECHA_CORTE"] = pd.to_datetime(
    df["FECHA_CORTE"].astype("string").str.strip(),
    format="%Y%m%d",
    errors="coerce"
)

print(df["FECHA_CORTE"].dtype)
display(df[["FECHA_CORTE"]].head())
```

```python
# Convertir PERIODO a un periodo mensual
df["PERIODO"] = pd.to_datetime(
    df["PERIODO"].astype("string").str.strip(),
    format="%Y%m",
    errors="coerce"
).dt.to_period("M")

# Verificar la conversión
print("Tipo de dato:", df["PERIODO"].dtype)
display(df[["PERIODO"]].head())
```
